#Loops

```python
for i in range(1,5,1):
  print(i)
```

```python
# Print all the numbers from 2 to 10 (both inclusive)

for i in range(2,11,1):
  print(i)
```

```python
for i in range(1,11,1):
  print(i,end = " ") #'end' character is used to end the line as per your wish
```

```python
for i in range(1,5,2):
  print("Suyash")
```

```python
# Print 1 to 5 (both inclusive) using only one loop command

for i in range(5):
  print(i+1)

```

```python
# Print 10 to 1 (both inclusive)

for i in range(0,10,1):
  print(10-i)

  #OR
print(end="\n")

for i in range(10,0,-1):
  print(i)
```

```python
for i in "Suyash":
  print(i)
```

#Practice Problems

```python
# Print all the even numbers between 0 to 10 (both inclusive)

for i in range(0,11,2):
  print(i)
```

```python
# Print all the even numbers between 0 to 10 (both inclusive) using only one parameter.

for i in range(6):
  print(i*2)

  #OR
print(end= "\n")

for i in range(11):
  if i%2 ==0 :
    print(i)
```

```python
# Print all the odd numbers between 0 to 10 (both inclusive) using only one parameter.

for i in range(11):
  if i%2 != 0:
    print(i ,end=" ")

#OR
print(end="\n")
print(end="\n")

for i in range(1,11,2):
  print(i,end=" ")
```

```python
# Given a string, print all its character separated by "-"
# Suyash --> S-u-y-a-s-h

for i in "Suyas":
  print(i, end="-")
print("h")
```

```python
# Print the squares of all the int from 2 to 7 (both inclusive)

for i in range(2,8,1):
  print(i**2)
```

```python
#Print the  multiplication table of 5 for its first 10 multiples.

for i in range(1,11):
  print("5 * ",i," = ", 5*i)

```
