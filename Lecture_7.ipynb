```python
# Print all the odd numbers between 0 to 10 using while loop.
i = 1
while i<10:
  print(i)
  i = i+2

                    # OR
print(end= "\n")

i = 0
while i<11:
  if (i % 2 != 0):
    print (i)
  i = i+1
```

```python
# Print the sum of all the numbers between 1 to 10 (both inclusive)

# 1. For Loop
sum = 0
for i in range (1,11):
  sum += i
  print (sum)



# 2. While Loop
sum = 0
i = 1
while i <= 10:
  sum += i
  i += 1
  print(sum)
```
