```python
#Calculator --->
a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
print("Addition of the given two numbers is:", a+b)
print("Subtraction of the given two numbers is:", a-b)
print("Multiplication of the given two numbers is:", a*b)
print("Division of the given two numbers is:", a/b)
print("Floored division of the given two numbers is:", a//b)
print("Power of the given two numbers is:",a**b)
print("Remainder of the given two numbers is:",a%b)
```
