```python
# Take marks obtained from the user ,
# if marks < 30 --> print "Low"
# if marks is between 30 and 70 --> print "Average"
# if marks is above 70 --> print "High"

# print using if-else statement

a = int(input("Enter the marks obtained:"))
if a < 30:
  print("Low")
else:
  if a>=30 and a<=70:
    print("Average")
  else:
    print("High")


# print using elif statement

a= int(input("Enter the marks obtained:"))
if a < 30:
  print("Low")
elif a >=30 and a <= 70:
  print("Average")
else:
  print("High")
```
