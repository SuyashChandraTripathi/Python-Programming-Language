# List Slicing



```python
a =[4 ,6 ,7, 9, 2, 1]

print(a)

print(a[1:5])

print(a[:4])

print(a[2:])

print(a[:])

print(a[1:4:2])

print(a[::2])
```

#Strings

```python
# Print the given string in reverse order --> "Suyash"

name = "Suyash"
for i in reversed(name):
  print(i , end = " ")

# OR
print("\n")

name = "Suyash"
n = len(name)
for i in range(n):
  print(name[n-1-i] , end = " ")

# OR
print("\n")

name = "Suyash"
print(name[::-1] , end = " ")

```
