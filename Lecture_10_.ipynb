#Lists & Indexing

```python
#Write a code to print the elements present at even indexes.
marks = [90 , 86 , 97 , 83 , 70]
for i in range(len(marks)):
  if i % 2 == 0 :
    print(marks[i])
```

```python
# Write a code to find the sum of all the elements of the list.
sum = 0
marks = [60 ,70 ,80 ,90]
print ("The list is: " , end = "")
for i in range(len(marks)):
  print(marks[i] , end = " ")
for j in marks:
  sum = sum + j
print ("\nThe sum of elements in the list is: " , sum)

```

```python
# Write a code to find the average of all the elements of the lists.
sum = 0

marks = [60,70, 80, 90]
print("The list is: " , end = "" )
print(marks , end= " ")
for i in marks:
  sum += i
ele = len(marks)
print ("\nAverage of the elements in the list is: " , sum/ele)
```
