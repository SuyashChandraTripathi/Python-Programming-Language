#Nested Loops


```python
# Print the following pattern using nested loop.
# * * * * *
# * * * * *
# * * * * *
# * * * * *
a = int(input("Enter the number of rows: "))
b = int(input("Enter the number of starts in each row: "))
for i in range(0 , a , 1):
  for j in range(0 , b , 1):
    print ("*" , end = " ")
  print (end = "\n")
```
