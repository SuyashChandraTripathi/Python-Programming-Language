```python
# Write a program to identify whether a number is positive, negative or zero.

a = float(input("Enter a number:"))
if a > 0:
  print("This is a positive number")
elif a==0:
  print("The number is a zero")
else:
  print("This is a negative number")
```

```python
# Take two numbers and print the larger number.
# If both are equal, print "Both the numbers are equal"

a = float(input("Enter the first number:"))
b = float(input("Enter the second number:"))
if a>b:
  print(a," is greater than ",b)
elif b>a:
  print(b," is greater than ",a)
else:
  print(a ," and ",b," are same numbers")
```

```python
# Assign grades based upon marks.

# 90 and above --> Grade A
# 75-89 --> Grade B
# 60-74 --> Grade C
# 40-59 --> Grade D
# Below 40 --> Fail

marks = int(input("Enter your obtained marks:"))
if marks >= 90:
  print("Grade A")
elif marks >=75 and marks <= 89 :
  print("Grade B")
elif marks >=60 and marks <= 74 :
  print("Grade C")
elif marks >= 40 and marks <= 59 :
  print("Grade D")
elif marks < 40 :
  print("Beta, tumse na ho paayega !!")
```

```python
# Body Mass Index (BMI) is a simple measure that helps determine
# whether a person's weight is appropriate for their height.

# Write a python program that takes the BMI value as input from the user
# and displays the corresponding weight category using the following rules:

# BMI value category

# Less than 18.5 --> Underweight
# 18.5 to 24.9 --> Normal
# 25.0 to 29.9 --> Overweight
# 30.0 or above --> Obese

a = float(input("Enter your BMI:"))
if a < 18.5:
  print("Underweight")
elif a >= 18.5 and a <= 24.9:
  print("Normal")
elif a >= 25.0 and a <= 29.9:
  print("Overweight")
else:
  print("Obese")
```
