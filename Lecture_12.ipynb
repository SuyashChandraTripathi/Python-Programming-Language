#Functions

```python
# Write a function that prints "Hello name" for whatever name you want.

a = str(input("Enter you name: "))
def hello():
  print("Hello" , a)
hello()

# OR
print("\n")

def hello(name):
  print("Hello" , name)
hello("Suyash Chandra Tripathi")
```

##Positional Arguments


```python
def show (name , age):
  print ("Given name is" , name)
  print ("Given age is" , age)
show("Suyash" , 19)

# But if you reverse it
print("\n")

show(19 , "Suyash")
```

##Keyworded Arguments


```python
def show(name , age):
  print("Given name is" , name)
  print("Given age is" , age)
show(name = "Suyash" , age = 19)

# OR
print("\n")

show(age = 19 , name = "Suyash")
```
